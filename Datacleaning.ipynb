```python
Anime_df['english_title'] = Anime_df['english_title'].fillna(Anime_df['title'])
```

```python
Anime_df['english_title'].isnull().sum()
```

```python
Anime_df['genres'] = Anime_df['genres'].fillna('Unknown')
```

```python
Anime_df['genres'].isnull().sum()
```

```python
Anime_df['rating_anilist'] = Anime_df['rating_anilist'].fillna(Anime_df['rating_anilist'].median())
Anime_df['rating_mal'] = Anime_df['rating_mal'].fillna(Anime_df['rating_mal'].median())
Anime_df['episodes'] = Anime_df['episodes'].fillna(1)
Anime_df['format'] = Anime_df['format'].fillna("unavailable")
```

```python
Anime_df = Anime_df.dropna(subset=['season'])
```

```python
Anime_df = Anime_df.dropna(subset=['year'])
```

```python
Anime_df = Anime_df.dropna(subset=['synopsis'])
```

```python
Anime_df['studio'] = Anime_df['studio'].fillna("unknown")
```

```python
Anime_df['source_material'] = Anime_df['source_material'].fillna("unknown")
```

```python
Anime_df = Anime_df.dropna(subset=['mal_members'])
```

```python
max_rank = Anime_df['mal_rank'].max()
fill_value = max_rank + 1

Anime_df['mal_rank'] = Anime_df['mal_rank'].fillna(fill_value)
```

```python
Anime_df['start_date'] = Anime_df['start_date'].fillna("Unknown")
Anime_df['end_date'] = Anime_df['end_date'].fillna("Ongoing")
```

```python
mode_duration = Anime_df['duration_per_ep'].mode()[0]
Anime_df['duration_per_ep'] = Anime_df['duration_per_ep'].fillna(mode_duration)
```
